# **The Vitamin Supplements Problem**

Importing the pulp package below

```python
# !pip show  pulp
# !pip install pulp

import pulp as pp
```

Initializing the model below

```python
model = pp.LpProblem(name='vitamin-supplements',sense=pp.LpMinimize)
```

Declaring the variables. The code that is commented out can be used instead of the active code if fractional tablets can be taken. I am assuming that only whole tablets can be taken, so the category is set to integer below.

```python
#SV = pp.LpVariable("SuperVit", lowBound=0)
#NH = pp.LpVariable("NewHealth", lowBound=0)

SV = pp.LpVariable("SuperVit", lowBound=0,cat="Integer")
NH = pp.LpVariable("NewHealth", lowBound=0,cat="Integer")
```

Building the model by setting the parameters of the price and tablets information below, with the daily criteria of the vitamins

```python
#Price
model += 0.20 * SV + 0.30 * NH

#Vitamin C
model += 20 * SV + 30 * NH >= 60

#Calcium
model += 500 * SV + 250 * NH >= 1000

#Iron
model += 9 * SV + 2 * NH >= 18

#Niacin
model += 2 * SV + 10 * NH >= 20

#Magnesium
model += 60 * SV + 90 * NH >= 360
```

Solving the model and printing the results

```python
model.solve(pp.PULP_CBC_CMD(msg=0))

print("SuperVit tablets:", pp.value(SV))
print("NewHealth tablets:", pp.value(NH))
print("Minimum cost: $", round(pp.value(model.objective),ndigits=2))
```
